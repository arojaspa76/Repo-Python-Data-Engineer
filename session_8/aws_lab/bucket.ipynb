```python
import boto3
```

## Creacion de un bucket en aws usando python

```python
s3 = boto3.client('s3', region_name='us-east-1')
s3.create_bucket(Bucket='bsg-python-data-engineering-bucket')
```

### Carga de archivo csv al bucket de S3 y en carpeta datasets

```python
s3.upload_file('data.csv', 'bsg-python-data-engineering-bucket', 'datasets/data.csv')
```

### Carga de archivo csv al bucket de S3 y en carpeta datasets

```python
s3.upload_file('patients.csv', 'bsg-python-data-engineering-bucket', 'datasets/patients.csv')
```
